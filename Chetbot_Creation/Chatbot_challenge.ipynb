#2.Initialize Chatbot

## 2.1 Import Required Libraries

```python
from langchain.memory import ConversationBufferMemory
from langchain_ollama.llms import OllamaLLM
from langchain_core.prompts import PromptTemplate
from langchain.chains import LLMChain
```

## 2.2 Configure Loan Details

```python
loan_details = {
    "name": "Rahul Sharma",
    "loan_type": "Home Loan",
    "days": "45",
    "amount": "₹32,500"
}
```

## 2.3 Initialize chatbot memory and setup LLM

```python
initial_query=f"Hello {loan_details['name']}. I am calling from ICICI Home Finance regarding payment. Can I continue?"
memory = ConversationBufferMemory(memory_key="chat_history")
memory.chat_memory.add_ai_message(
    initial_query
)
llm = OllamaLLM(model="mistral")
```

## 2.4 Create Conversation Template and Chain

```python
template = f"""You are a professional loan collection agent for ICICI Bank. Start the conversation EXACTLY as follows:"""+initial_query+"""
Customer details: """ + ", ".join(f"{key}: {value}" for key, value in loan_details.items()) + """
Guidelines for the conversation:

Dispute:
○ Customer claims the loan may not belong to them, or that the loan has already
been closed.
Generic Yes:
○ Customer expresses interest in speaking and requests the payment date.
○ Once confirmed, conclude the conversation with a thank you message.
Generic No:
○ Customer declines the conversation.
○ Politely persuade them by mentioning the importance of discussing loan
repayment (repeat 2-3 times).
○ If the customer consistently declines, end the conversation.
Wrong Number:
○ If the customer states it is the wrong number, apologize for the inconvenience.
○ Wish them a good day ahead.
In short:
If Dispute then: Response = Please contact your nearest branch or our customer care. Thank you (Like Not his loan account, Loan already closed, etc)
If:generic_yes: Your <Loan Type> payment is pending for <> days/month. You have to pay <>. When will you pay. -----> Get the payment Confirmation and thank him

If:generic_no: Persuade customer to agree to talk (Atleast 3 times) ------>
Thank you for your time

If:wrong_no: Apologies for the inconvenience. Have a nice day.

Previous conversation:
{chat_history}

New human question: {question}
Response:"""
prompt = PromptTemplate.from_template(template)
# Create the conversation chain
conversation = LLMChain(
    llm=llm,
    prompt=prompt,
    verbose=True,
    memory=memory
)
```

## 3.Run Chatbot

```python
# List of keywords indicating conversation should end
closing_keywords=[]
ollama_thread = threading.Thread(target=run_ollama)
ollama_thread.start()
time.sleep(5)  # Wait for server initialization
# Continue the conversation
print("Bot: "+initial_query)
while True:
    user_input = input("You: ").strip()

    # Manual exit commands
    if user_input.lower() in ["exit", "quit", "bye"]:
        print("Bot: Thank you for your time. Have a great day!")
        break

    # Generate response
    response = conversation.run(question=user_input)
    print(f"Bot: {response}")

    # Auto-close detection
    if any(keyword in response.lower() for keyword in closing_keywords):
        break  # End conversation if closing phrase detected
```
